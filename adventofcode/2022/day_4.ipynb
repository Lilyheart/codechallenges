# Helper Function

```python
def process_raw_input(raw_pair_list: list[str]) -> list[list[dict[str, str]]]:
	return [[{"start": int(elf.split("-")[0]), "end": int(elf.split("-")[1])} for elf in line.strip().split(",")] for line in raw_pair_list.split("\n")]
```

```python
def is_consumed(elf_pair: list[dict[str, str]]) -> bool:
	if elf_pair[0]["start"] >= elf_pair[1]["start"] and elf_pair[0]["end"] <= elf_pair[1]["end"]:
		return True
	if elf_pair[1]["start"] >= elf_pair[0]["start"] and elf_pair[1]["end"] <= elf_pair[0]["end"]:
		return True
	return False
```

```python
def get_total_consumed(elf_pairs: list[list[dict[str, str]]]) -> int:
	total_consumed = 0
	for elf_pair in elf_pairs:
		if is_consumed(elf_pair):
			total_consumed += 1

	return total_consumed
```

```python
def is_overlapped(elf_pair: list[dict[str, str]]) -> bool:
	for section in range(elf_pair[0]["start"], elf_pair[0]["end"] + 1):
		if section in range(elf_pair[1]["start"], elf_pair[1]["end"] + 1):
			return True
	return False
```

```python
def get_total_overlapped(elf_pairs: list[list[dict[str, str]]]) -> int:
	total_overlapped = 0
	for elf_pair in elf_pairs:
		if is_overlapped(elf_pair):
			total_overlapped += 1

	return total_overlapped
```

# Unit Tests

```python
import unittest

class TestNotebook(unittest.TestCase):
	test_input = """2-4,6-8
2-3,4-5
5-7,7-9
2-8,3-7
6-6,4-6
2-6,4-8"""

	def setUp(self):
		self.pair_list = process_raw_input(self.test_input)

	def test_is_consumed(self):
		self.assertEqual(is_consumed(self.pair_list[0]), False)
		self.assertEqual(is_consumed(self.pair_list[1]), False)
		self.assertEqual(is_consumed(self.pair_list[2]), False)
		self.assertEqual(is_consumed(self.pair_list[3]), True)
		self.assertEqual(is_consumed(self.pair_list[4]), True)
		self.assertEqual(is_consumed(self.pair_list[5]), False)

	def test_get_total_consumed(self):
		self.assertEqual(get_total_consumed(self.pair_list), 2)

	def test_is_overlapped(self):
		self.assertEqual(is_overlapped(self.pair_list[0]), False)
		self.assertEqual(is_overlapped(self.pair_list[1]), False)
		self.assertEqual(is_overlapped(self.pair_list[2]), True)
		self.assertEqual(is_overlapped(self.pair_list[3]), True)
		self.assertEqual(is_overlapped(self.pair_list[4]), True)
		self.assertEqual(is_overlapped(self.pair_list[5]), True)
		self.assertEqual(is_overlapped([self.pair_list[5][1], self.pair_list[5][0]]), True)

	def test_get_total_overlapped(self):
		self.assertEqual(get_total_overlapped(self.pair_list), 4)

unittest.main(argv=[''], verbosity=2, exit=False)
```
