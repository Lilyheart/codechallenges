```python
# Using readlines()
input = open("input.txt", "r")
# lines = input.readlines() # has line returns
lines = input.read().splitlines() # doesn't have line returns

thisCalories = 0
maxCalories = 0
for line in lines:
	if len(line) == 0:
		if thisCalories > maxCalories:
			maxCalories = thisCalories
		thisCalories = 0
	else:
		thisCalories += int(line)

print(maxCalories)
```
