# Helper Function

```python
def process_raw_input(raw_calories: str) -> list[list[str]]:
	"""
	Read the raw text file input and convert into a list of lists containing calories for each elf
	"""
	elves = raw_calories.split("\n\n")
	calories = [list(filter(None, elf.split("\n"))) for elf in elves]
	calories_sum = [sum(int(item) for item in calories) for calories in calories]

	return calories_sum
```

```python
def max_calories(calories_by_elf: list[int], count=1) -> int:
	"""
	Returns the sum of calories for the top `count` number of elves determined by the count argument that defaults to 1.
	"""
	return sum(sorted(calories_by_elf, reverse=True)[:count])
```

# Unit Tests

```python
import unittest

class TestNotebook(unittest.TestCase):
	test_input = """1000
2000
3000

4000

5000
6000

7000
8000
9000

10000"""
	
	def test_process_raw_input(self):
		expected = [6000, 4000, 11000, 24000, 10000]
		self.assertEqual(process_raw_input(self.test_input), expected)
	
	def test_max_calories_1(self):
		self.assertEqual(max_calories(process_raw_input(self.test_input)), 24000)
	
	def test_max_calories_3(self):
		self.assertEqual(max_calories(process_raw_input(self.test_input), 3), 45000)

unittest.main(argv=[''], verbosity=2, exit=False)
```

# Puzzle Solve

```python
import os

with open("input" + os.sep + "1.txt", "r", encoding="utf-8") as input_values:
	raw_calories = input_values.read()

calories = process_raw_input(raw_calories)
top1 = max_calories(calories)
top3 = max_calories(calories, 3)

print(f"top1: {top1}")
print(f"top3: {top3}")
```
