# Helper Function

```python
def process_raw_input(raw_strategy_guide: list[str]):
	return [line.strip() for line in raw_strategy_guide.split("\n")]
```

```python
def find_duplicates(contents: list[str]) -> str:
	# Combine contents keeping any correct duplicate items.
	actual = []
	goal = ""
	for section in contents:
		actual += list(set(section))
		goal += section
	goal = list(set(goal))

	results = list(number for number in actual if actual.count(number) == len(contents))

	if len(results) == 0:
		return None
	elif len(results) == len(contents):
		return results[0]
	else:
		raise Exception("Extra Duplicates", results)
```

```python
def get_item_priority(item: str) -> int:
	if ord(item) >= 97:
		return ord(item) - 96
	else:
		return ord(item) - 38
```

```python
def get_item_priority_sum(list_of_contents: list[list[str]]) -> int:
	sum = 0
	for contents in list_of_contents:
		sum += get_item_priority(find_duplicates([contents[0:len(contents)//2], contents[len(contents)//2:]]))

	return sum
```

```python
def get_badge_priority_sum(list_of_contents: list[list[str]]) -> int:
	sum = 0
	for i in range(0, len(list_of_contents), 3):
		sum += get_item_priority(find_duplicates(list_of_contents[i:i+3]))

	return sum
```

# Unit Tests

```python
import unittest

class TestNotebook(unittest.TestCase):
	test_input = """vJrwpWtwJgWrhcsFMMfFFhFp
jqHRNqRjqzjGDLGLrsFMfFZSrLrFZsSL
PmmdzqPrVvPwwTWBwg
wMqvLMZHhHMvwLHjbvcjnnSBnvTQFn
ttgJtRGJQctTZtZT
CrZsJsPPZsGzwwsLwLmpwMDw"""

	def setUp(self):
		self.list_of_contents = process_raw_input(self.test_input)

	def test_find_duplicate(self):
		self.list_of_contents = [[line[0:len(line)//2], line[len(line)//2:]] for line in self.list_of_contents]
		self.assertEqual(find_duplicates(self.list_of_contents[0]), "p")
		self.assertEqual(find_duplicates(self.list_of_contents[1]), "L")
		self.assertEqual(find_duplicates(self.list_of_contents[2]), "P")
		self.assertEqual(find_duplicates(self.list_of_contents[3]), "v")
		self.assertEqual(find_duplicates(self.list_of_contents[4]), "t")
		self.assertEqual(find_duplicates(self.list_of_contents[5]), "s")

	def test_get_item_priority(self):
		self.assertEqual(get_item_priority("a"), 1)
		self.assertEqual(get_item_priority("z"), 26)
		self.assertEqual(get_item_priority("A"), 27)
		self.assertEqual(get_item_priority("Z"), 52)

	def test_get_item_priority_sum(self):
		self.assertEqual(get_item_priority_sum(self.list_of_contents), 157)

	def test_get_badge_priority_sum(self):
		self.assertEqual(get_badge_priority_sum(self.list_of_contents), 70)

unittest.main(argv=[''], verbosity=2, exit=False)
```
