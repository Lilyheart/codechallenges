```python
# Using readlines()
input = open("input.txt", "r")
# lines = input.readlines() # has line returns
lines = input.read().splitlines() # doesn't have line returns

thisCalories = 0
elvesCalories = []
for line in lines:
	if len(line) == 0:
		elvesCalories.append(thisCalories)
		thisCalories = 0
	else:
		thisCalories += int(line)

print(sum(sorted(elvesCalories)[-3:]))
```
