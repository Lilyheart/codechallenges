```python
def process_raw_input(raw_pair_list: list[str]) -> list[list[dict[str, str]]]:
	return [[{"start": int(elf.split("-")[0]), "end": int(elf.split("-")[1])} for elf in line.strip().split(",")] for line in raw_pair_list.split("\n")]
```
