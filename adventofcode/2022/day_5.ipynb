# Helper Function

```python
def process_raw_input(raw_input: list[str]) -> list[ dict[list[str]], list[dict[int]] ]:
	raw_stacks, raw_moves = raw_input.split("\n\n")
	stacks = {x: [] for x in range(1, int(max(raw_stacks.split("\n")[-1].split( ))) + 1)}
	for row in raw_stacks.split("\n"):
		if "[" not in row:
			continue
		for i in range(1, len(stacks) + 1):
			if len(row[4*i-3].strip()) == 1:
				stacks[i].append(row[4*i-3].strip())

	moves = [{"move": int(moves.split(" ")[1]), "from": int(moves.split(" ")[3]), "to": int(moves.split(" ")[5])} for moves in raw_moves.split("\n")]
	return stacks, moves
```

```python
def move_by_block(stacks: dict[list[str]], move: dict[int]) -> dict[str]:
	for i in range(move["move"]):
		block = stacks[move["from"]].pop(0)
		stacks[move["to"]].insert(0, block)
	return stacks
```

```python
def move_by_stack(stacks: dict[list[str]], move: dict[int]) -> dict[str]:
	to_move = stacks[move["from"]][:move["move"]]
	to_keep = stacks[move["from"]][move["move"]:]
	stacks[move["from"]] = to_keep
	stacks[move["to"]] = to_move + stacks[move["to"]]
	return stacks
```

```python
def get_stack_string(stacks: list[str]) -> str:
	result = ""
	for stack in stacks:
		if len(stacks[stack]) > 0:
			result += stacks[stack][0]
	return result
	# return "".join([stacks[stack][0] for stack in stacks])
```

```python
def process_by_block_moves(stacks: dict[list[str]], moves: dict[int]) -> dict[list[str]]:
	for move in moves:
		move_by_block(stacks, move)
	return stacks
```

```python
def process_by_stack_moves(stacks: dict[list[str]], moves: dict[int]) -> dict[list[str]]:
	for move in moves:
		move_by_stack(stacks, move)
	return stacks
```

# Unit Tests

```python
import unittest

class TestNotebook(unittest.TestCase):
	test_input = """    [D]    
[N] [C]    
[Z] [M] [P]
 1   2   3 

move 1 from 2 to 1
move 3 from 1 to 3
move 2 from 2 to 1
move 1 from 1 to 2"""

	def setUp(self):
		self.stacks, self.moves = process_raw_input(self.test_input)

	def test_get_stack_string(self):
		self.assertEqual(get_stack_string(self.stacks), "NDP")

	def test_move_by_block(self):
		self.assertEqual(get_stack_string(self.stacks), "NDP")
		move_by_block(self.stacks, self.moves[0])
		self.assertEqual(get_stack_string(self.stacks), "DCP")
		move_by_block(self.stacks, self.moves[1])
		self.assertEqual(get_stack_string(self.stacks), "CZ")
		move_by_block(self.stacks, self.moves[2])
		self.assertEqual(get_stack_string(self.stacks), "MZ")
		move_by_block(self.stacks, self.moves[3])
		self.assertEqual(get_stack_string(self.stacks), "CMZ")

	def test_process_by_block_moves(self):
		self.assertEqual(get_stack_string(process_by_block_moves(self.stacks, self.moves)), "CMZ")

	def test_move_by_stack(self):
		self.assertEqual(get_stack_string(self.stacks), "NDP")
		move_by_stack(self.stacks, self.moves[0])
		self.assertEqual(get_stack_string(self.stacks), "DCP")
		move_by_stack(self.stacks, self.moves[1])
		self.assertEqual(get_stack_string(self.stacks), "CD")
		move_by_stack(self.stacks, self.moves[2])
		self.assertEqual(get_stack_string(self.stacks), "CD")
		move_by_stack(self.stacks, self.moves[3])
		self.assertEqual(get_stack_string(self.stacks), "MCD")

	def test_process_by_stack_moves(self):
		self.assertEqual(get_stack_string(process_by_stack_moves(self.stacks, self.moves)), "MCD")

unittest.main(argv=[''], verbosity=2, exit=False)
```
