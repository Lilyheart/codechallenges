# Helper Function

```python
from enum import Enum, auto
class ScoreType(Enum):
	BYMOVE = auto()
	BYSCORE = auto()
```

```python
def process_raw_input(raw_strategy_guide: list[str]):
	return [line.strip().split(" ") for line in raw_strategy_guide.split("\n")]
```

```python
def score_round(moves: list[str], score_type: ScoreType) -> int:
	"""
	Returns the score of a move based on if the players move is based on move or score goal.
	"""
	OPP_SHAPE = {"A": "Rock", "B": "Paper", "C": "Scissors"}
	ME_SHAPE = {"X": "Rock", "Y": "Paper", "Z": "Scissors"}
	ME_GOAL = {"X": "Lost", "Y": "Draw", "Z": "Win"}

	RESULTS = {"PR": "Lost", "SP": "Lost", "RS": "Lost", "SR": "Win", "PS": "Win", "RP": "Win", "RR": "Draw", "PP": "Draw", "SS": "Draw"}

	match score_type:
		case ScoreType.BYMOVE:
			result = RESULTS[OPP_SHAPE[moves[0]][0] + ME_SHAPE[moves[1]][0]]
			shape = ME_SHAPE[moves[1]]
		case ScoreType.BYSCORE:
			result = ME_GOAL[moves[1]]
			shapeInitial = {game[0]: game[1] for game, subresult in RESULTS.items() if subresult == result}[OPP_SHAPE[moves[0]][0]]
			shape = [ME_SHAPE[shape] for shape in ME_SHAPE if ME_SHAPE[shape][0] == shapeInitial][0]

	SHAPE_SCORE = {"Rock": 1, "Paper": 2, "Scissors": 3}
	WIN_SCORE = {"Lost": 0, "Draw": 3, "Win": 6}

	return SHAPE_SCORE[shape] + WIN_SCORE[result]
```

```python
def get_total_score(strategy_guide: list[list[str]], score_type: ScoreType) -> int:
	"""
	Returns the total score for a game based on if the players move is based on the move or score goal.
	"""
	score = 0
	for move in strategy_guide:
		score += score_round(move, score_type)

	return score
```

# Unit Tests

```python
import unittest

class TestNotebook(unittest.TestCase):
	test_input = """A Y
B X
C Z"""

	def setUp(self):
		self.strategy_guide = process_raw_input(self.test_input)

	def test_score_by_moves_1(self):
		self.assertEqual(score_round(self.strategy_guide[0], ScoreType.BYMOVE), 8)
	
	def test_score_by_moves_2(self):
		self.assertEqual(score_round(self.strategy_guide[1], ScoreType.BYMOVE), 1)

	def test_score_by_moves_3(self):
		self.assertEqual(score_round(self.strategy_guide[2], ScoreType.BYMOVE), 6)

	def test_score_by_score_1(self):
		self.assertEqual(score_round(self.strategy_guide[0], ScoreType.BYSCORE), 4)

	def test_score_by_score_2(self):
		self.assertEqual(score_round(self.strategy_guide[1], ScoreType.BYSCORE), 1)

	def test_score_by_score_3(self):
		self.assertEqual(score_round(self.strategy_guide[2], ScoreType.BYSCORE), 7)

	def test_get_total_score_by_moves(self):
		self.assertEqual(get_total_score(self.strategy_guide, ScoreType.BYMOVE), 15)

	def test_get_total_score_by_score(self):
		self.assertEqual(get_total_score(self.strategy_guide, ScoreType.BYSCORE), 12)

unittest.main(argv=[''], verbosity=2, exit=False)
```
